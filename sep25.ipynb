## Recap: The Brown Corpus

The **Brown corpus** (Francis and Kučera, 1964) is a corpus of American English text in several **genres**.


```python
from nltk.corpus import brown
from nltk.probability import FreqDist

brown_news = brown.words(categories="news")
brown_fiction = brown.words(categories="fiction")

brown_news_dist = FreqDist(brown_news)
brown_fiction_dist = FreqDist(brown_fiction)
```

## Loops

A **loop** is a Python construct that allows you to run a piece of code repeatedly.

When we write a loop, we usually **loop through** (or "loop over," or "iterate through," or "iterate over") a `list`, `set`, `tuple`, `dict`, `str`, etc., and run the repeated code for each item.

## Example: Loop Through Ints

```python
# A "range" is a list-like object containing ints
list(range(10))
```

```python
# Loop through a range and do something to each int
for j in range(10):
    print(i * 2)
```

## Example: Loop Through Pairs

```python
# "zip" lets you create a list of pairs
list(zip(range(5), "abcde", "vwxyz"))
```

```python
for k, v, w in zip(range(5), "abcde", "vwxyz"):
    print(k, v, w, sep=": ")
```

```python
# "enumerate(x)" means "zip(range(len(x)), x)"
for k, v in enumerate("abcde"):
    print(k, v, sep=": ")
```

## Example: Loop Through A Set

Remember that `set`s are **unordered**, so we don't know what order a `set` will be looped through.

```python
for i in {1, 1, 2, 3, 5}:
    print(i)
```

```python
for i in {5, 3, 2, 1, 1}:
    print(i)
```

## Example: Loop Through A Dict

```python
x = {"a": 0, "b": 1, "c": 2, "d": 3, "e": 4}
```

```python
# Loop through the keys of the dict
for k in x:
    print(k, x[k], sep=": ")
```

```python
# Loop through the items of the dict
for k, v in x.items():
    print(k, v, sep=": ")
```

## Example: Visualizing a FreqDist

Let's use a loop to visualize the most common words in a `FreqDist` in a prettier format.

**Note:** `\t` is a **tab character** (try asking [TerrierGPT](https://terriergpt.bu.edu/share/W6TDqQbnFnugBlFg1U9kU) what that means).

```python
for word, count in brown_news_dist.most_common(5):
    print("The token type \"" + word + "\"\toccurs " + str(count) + "\ttimes.")
```

## Conditionals

**Conditionals** (a.k.a. "if statements") are a Python construct that lets you execute code contingent upon a certain condition.

```python
# Example: Print large ints
for i in range(10):
    if i <= 5:
        print(i)
```

## Else and Elif ("Else If")

```python
# Example: letter grades
for i in range(90, 101):
    if i < 93:
        print(i, "A-", sep=":\t")
    elif i > 97:
        print(i, "A+", sep=":\t")
    else:
        print(i, "A", sep=":\t")
```

## Exercise: Order of Condition Evaluation

```python
# This code does the same thing
for i in range(90, 101):
    if i < 93:
        print(i, "A-", sep=":\t")
    elif i < 98:  # Why don't we have to say "i >= 93 and i < 98"?
        print(i, "A", sep=":\t")
    else:
        print(i, "A+", sep=":\t")
```

## Exercise: "Elsewhere Condition"

The following code does the same thing, but is dispreferred. Why?

```python
for i in range(90, 101):
    if i < 93:
        print(i, "A-", sep=":\t")
    elif i < 98:
        print(i, "A", sep=":\t")
    elif i <= 100:  # No "else"
        print(i, "A+", sep=":\t")
    else:
        print(i + " is not a valid grade")
```

## Conditions Are Bools (Truth Values)

```python
for i in range(4):
    print(i, i > 2, type(i > 2), sep="\t")
```
