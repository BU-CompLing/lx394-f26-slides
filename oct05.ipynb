For example, the first 7 Fibonacci numbers are 1, 1, 2, 3, 5, 8, 13.

How do we find the $n$th Fibonacci number, $\varphi_n$?

```python
def fibonacci(n: int) -> int:
    pass
```

## Solution Using a For Loop

This code works, but is very ugly.

```python
def fibonacci(n: int) -> int:
    fibs = [1, 1]
    for i in range(2, n + 1):
        fibs.append(fibs[-1] + fibs[-2])
    return fibs[-1]
```

```python
[fibonacci(i) for i in range(50, 150, 20)]
```

## Solution Using Recursion

**Recursion** is when a function calls itself.

```python
def fib(n: int) -> int:
    if n < 2:
        return 1
    return fib(n - 1) + fib(n - 2)
```

## Risks of Recursion

What's wrong with the following code?

```python
def fib(n: int) -> int:
    return fib(n - 1) + fib(n - 2)
```

## Recursion and Efficiency

The recursive version of `fib` is **extremely inefficient**. In fact:
- `fib(50)` takes at least 1 hour
- `fib(69)` takes at least 1 year
- `fib(117)` takes at least 13.8 billion years.

```python
def fib(n: int) -> int:
    if n < 2:
        return 1
    return fib(n - 1) + fib(n - 2)
```

Why? How can we make `fib` more efficient?

## Memoization

One way to make `fib` more efficient is to **save** ("cache") **previously computed outputs**.

This is called **memoization**.

We can implement memoization in Python by creating a **class**.

## Classes, Objects, and Data Types

**Classes** are custom types, which can be used to create objects.

```python
# Example
class HWSubmission:
    def __init__(self):
        # Class properties
        self.grade = 100 
        
    # Class methods
    def add_points(self, n: int):
        self.grade += n
        
    def remove_points(self, n: int):
        self.grade -= n
```

## Memoization Using a Class

```python
class Fibonacci:
    def __init__(self):
        self.cache = {0: 1, 1: 1}
        
    def fib(self, n: int) -> int:
        if n in self.cache:
            return self.cache[n]
        else:
            fib_n = self.fib(n - 1) + self.fib(n - 2)
            self.cache[n] = fib_n
            return fib_n
```

```python
fib = Fibonacci()
fib.fib(1300)
```
