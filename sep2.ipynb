## Dataset: Wikipedia and ChatGPT Corpus

We'll analyze a corpus of **Wikipedia articles** and **ChatGPT-generated articles** collected by [Singh et al. (2024)](https://www.sciencedirect.com/science/article/pii/S294971912300047X).

Wikipedia texts come from the introduction section of articles officially designated as "[Good Articles](https://en.wikipedia.org/wiki/Wikipedia:Good_articles)."

image

## ChatGPT Articles

ChatGPT articles were generated using the following prompt template:

```
Write a summary of 'Flag of Mauritania' in about 90 words
```

The final corpus contains…

- 22,081 Wikipedia articles
- 22,081 ChatGPT-generated articles

## Load Corpus

We'll start by running some code that will load the corpus. Later in the semester you will learn how this code works.

This code is written in a programming language called **Python**.

````python
# Load the Wikipedia and ChatGPT corpus
import pickle

with open("data/wiki_vs_chatgpt_articles.p", "rb") as f:
    dataset = pickle.load(f)
    wiki_articles = dataset["wiki"]
    chatgpt_articles = dataset["chatgpt"]
````

## Import NLTK

Python has many **packages** containing software that you can use in a program.

In this course we will be using the **Natural Language Toolkit** (NLTK) to process and analyze text. We load the NLTK package using an `import` statement.

````python
# We are loading the NLTK package.
# Don't worry about what this means for now!
from nltk.probability import FreqDist
````

## Searching for Words

NLTK's `.concordance` method lets us search for words within a text.

````python
# Search for the word "delve" in the Wikipedia articles
wiki_articles.concordance("delve")
````

````python
# Search for the word "delve" in the ChatGPT articles
chatgpt_articles.concordance("delve")
````
