## Mutability

`tuple`s are like `list`s, but they are **immutable**.

```python
# Lists are mutable
x = [1, 2, 3]
x[1] = 5
x
```

## Recap: Loops

A **loop** is a Python construct that allows you to run a piece of code repeatedly.

When we write a loop, we usually **loop through** (or "loop over," or "iterate through," or "iterate over") a `list`, `set`,  `dict`, `str`, etc., and run the repeated code for each item.

## Examples of Loops

```python
# Loop through a range and do something to each int
for i in range(5):
    print(i * 2)
```

```python
# Loop through the items of the dict
x = {"a": 0, "b": 1, "c": 2, "d": 3, "e": 4}
for k, v in x.items():
    print(k, v, sep=": ")
```

## Recap: Conditionals

**Conditionals** (a.k.a. "if statements") are a Python construct that lets you execute code contingent upon a certain condition.

```python
# Example: Print large ints
for i in range(10):
    if i > 5:
        print(i)
```

## Recap: Else and Elif ("Else If")

```python
for i in range(90, 101):
    if i < 93:
        print(i, "A-", sep=":\t")
    elif i < 98:  
            print(i, "A", sep=":\t")
        else:
            print(i, "A+", sep=":\t")
```

```python
# Another way of writing the same code
for i in range(90, 101):
    if i < 93:
        print(i, "A-", sep=":\t")
    else:
        if i < 98:  
            print(i, "A", sep=":\t")
        else:
            print(i, "A+", sep=":\t")
```

## Constructing a List Using a Loop

We often use loops to construct `list`s, `set`s, `dict`s, etc.


```python
# Construct the list [0, 2, 4, 6, 8]
x = []
for i in range(5):
    x += [i * 2]
x
```

```python
# Construct the dict {0: 0, 1: 2, 2: 4, 3: 6, 4: 8}
x = dict()
for i in range(5):
    x[i] = i * 2
x
```

## Efficiency

Constructing a `list` using a loop can be computationally inefficient.

```python
# Very inefficient
x = []
for i in range(5):
    x += [i * 2]  # Two new lists are created, x gets deleted
```

```python
# Somewhat efficient
x = []
for i in range(5):
    x.append(i * 2)  # x gets expanded, no new lists created
```

```python
# Very efficient
x = [i * 2 for i in range(5)]  # x is created using an optimized algorithm
```
