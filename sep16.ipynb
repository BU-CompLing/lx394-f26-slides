## Dicts

A `dict` (short for "dictionary") is like a `list`, but it's not necessarily indexed by `int`s.

In other programming languages, `dicts` are sometimes called **mappings** or **hashmaps**.

```python
# Create a dict
capitals = {"UK": "London", "France": "Paris", "Germany": "Berlin", "Germany": "Munich"}

# Retrieve dict items
print(capitals["UK"])
print(capitals["France"])
print(capitals["Germany"])
```

```python
# Add a dict item
capitals["Italy"] = "Rome"

print(capitals)
```

```python
# Length of a dict
len(capitals)
```

## List and Set Comprehension

In mathematics and semantics, what does this notation mean?

$$\lbrace w | w \in \texttt{capitals.keys()} \text{ and } \texttt{len}(w) > 3 \rbrace$$

```python
{w for w in 𝚌𝚊𝚙𝚒𝚝𝚊𝚕𝚜.𝚔𝚎𝚢𝚜() if len(w) > 3}
```

The variable $w$ is a _bound variable_. It represents an item in the list `capitals.keys()`. The set contains $w$ if and only if its length is greater than 3.

| w | len(w) > 3? |
|---|---|
| `UK` | `False` |
| `France` | `True` |
| `Germany` | `True` |
| `Italy` | `True` |

## Exercise: Case-Insensitive Vocab Size

How do we modify our `vocab_size` function to make it optionally case-insensitive?

Hint: Try `"Hello world!".lower()`

```python
def vocab_size(corpus, case_sensitive=True):
    """Calculates the vocab size of a corpus."""
    if case_sensitive:
        return len(set(corpus))
    else:
        return len({i.lower() for i in set(corpus)})
```

```python
print(vocab_size(chatgpt_articles, case_sensitive=True))
print(vocab_size(chatgpt_articles, case_sensitive=False))
```
