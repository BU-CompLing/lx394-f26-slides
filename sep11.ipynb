## Comments and Documentation

_Comments_ are pieces of code that do nothing. They are used to _explain_ what the code does.

Most comments begin with `#`.

A [**docstring**](https://peps.python.org/pep-0257/) ("documentation string") is a special type of comment, enclosed in `""" ... """`, that describes what a function does.

```python
# One-line docstring 
def frequency(corpus, word):
    """Calculates the relative frequency of a word within a corpus."""
    return corpus.count(word) / len(corpus)
```

```python
# More detailed docstring
def frequency(corpus, word):
    """
    Calculates the relative frequency of a word within a corpus.
    
    :param corpus: An NLTK Text containing a corpus
    :param word: A word to search for in the corpus
    
    :return: The proportion of words in corpus that match word
    """
    return corpus.count(word) / len(corpus)
```

## Optional Parameters

Parameters can be given a **default value**. When a parameter has a default value, it's customary to give its value as a **keyword argument** when calling the function.

```python
def frequency(corpus, word, percentage=False):
    """Calculates the relative frequency of a word within a corpus."""
    freq = corpus.count(word) / len(corpus)
    return freq * 100 if percentage else freq
```

```python
print(frequency(wiki_articles, "the"))
print(frequency(wiki_articles, "the", percentage=True))
```
