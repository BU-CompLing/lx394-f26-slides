## Recap: Bayesian Classification

Let $C$ be a set of **labels** ("classes"). For a text $x$, a Bayesian classifier generates a **predicted label** $\hat{y} \in C$ as follows:

$$\hat{y} = \underset{c \in C}{\text{argmax}} \log_2(\mathbb{P}[x \mathrel{|} c]) + \log_2(\mathbb{P}[c])$$

where
- $\log_2(\mathbb{P}[x \mathrel{|} c])$ is estimated using a language model 
- $\log_2(\mathbb{P}[c])$ is the proportion of the training corpus with label $c$

## Recap: Bayes's Theorem

The Bayesian classification score comes from **Bayes's theorem**:

$$\mathbb{P}[c \mathrel{|} x] = \frac{\mathbb{P}[x \mathrel{|} c] \mathbb{P}[c]}{\mathbb{P}[x]}$$

- $\mathbb{P}[c]$ is called the **prior probability** of $c$
- $\mathbb{P}[x \mathrel{|} c]$ is called the **likelihood** of $x$ given $c$
- $\mathbb{P}[c \mathrel{|} x]$ is called the **posterior probability** of $c$ given $x$

## Exercise: Interpreting Bayesian Classification

Let's look at Bayes's theorem applied to spam vs. ham.

$$
\mathbb{P}[\text{spam} \mathrel{|} x] = \frac{\mathbb{P}[x \mathrel{|} \text{spam}]\mathbb{P}[\text{spam}]}{\mathbb{P}[x]}
$$

- **Prior:** How common is spam, in general?
- **Likelihood:** If I were a spammer, would I write this email?

## Exercise: Phonological Perception

Let's say you are listening to a sound wave $s$ and want to decide whether it's phoneme $p$ or $q$.

$$
\mathbb{P}[p \mathrel{|} s] = \frac{\mathbb{P}[s \mathrel{|} p]\mathbb{P}[p]}{\mathbb{P}[s]}
$$

- **Prior:** Does my language contain the phoneme $p$?
- **Likelihood:** How would I pronounce the phoneme $p$?

## Exercise: Scalar Implicature (Pragmatics)

Consider the sentence: $s(2) = {}$"Romi had two slices of pizza for dinner last night." How many slices of pizza did she eat in total?

$$\mathbb{P}[n = 2 \mathrel{|} s(2)] = \frac{\mathbb{P}[s(2) \mathrel{|} n = 2] \mathbb{P}[n = 2]}{\mathbb{P}[s(2)]}$$

- **Prior:** Does 2 seem like a reasonable amount of pizza for Romi to eat?
- **Likelihood:** Does $s(2)$ seem like a reasonable thing for Romi to say, if she had eaten 2 slices of pizza?

## Naïve Bayes Classification

When the likelihood ($\mathbb{P}[x \mathrel{|} c]$) is estimated using a **unigram language model**, the classifier is called a **naïve Bayes classifier**.

- The unigram language model is also called a **bag of words model**
- The words are also called **features**
- "Naïve" means that the features are all **independent**
- In HW 3, the language model is **binary**: for each $w$, it only predicts whether an email contains $w$, and doesn't care _how many_ times $w$ occurs in the email

$$\mathbb{P}[w_1w_2 \mathrel{|} c] = \mathbb{P}[w_1 \mathrel{|} c]\mathbb{P}[w_2 \mathrel{|} c]$$
