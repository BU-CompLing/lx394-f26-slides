## Recap: Language Modeling

**Language modeling** is an NLP task where, given a sequence of tokens $x = x_1x_2 \dots x_l$ where $x_1 = \texttt{'<s>'}$, the **language model** must predict the next token, $y$.

Given input $x$, the language model's output is:

$$
\hat{y} = \begin{Bmatrix}
   \text{the:} & \mathbb{P}[\text{the} | x] \\
   \text{of:} & \mathbb{P}[\text{of} | x] \\
   \text{in:} & \mathbb{P}[\text{in} | x] \\
   \vdots & \vdots
\end{Bmatrix}
$$


## Recap: (Log) Probability of a Text

The **probability** that the sequence $y = y_1 y_2 \dots y_l$ will occur after $x$ is given by:
$$
\mathbb{P}[y | x] = \prod_{i = 1}^{l} \mathbb{P}[y_i \mathrel{|} xy_1 y_2 \dots y_{i - 1}].
$$

The **log probability** that the sequence $y = y_1 y_2 \dots y_l$ will occur after $x$ is given by:
$$
\log_2(\mathbb{P}[y \mathrel{|} x]) = \sum_{i = 1}^{l} \log_2(\mathbb{P}[y_i \mathrel{|} xy_1 y_2 \dots y_{i - 1}]).
$$

## Recap: Perplexity

We evaluate a language model on a test corpus $y$ via its **perplexity** (ppl):

\begin{align*}
\text{ppl} &= \frac{1}{\mathbb{P}[y \mathrel{|} \texttt{<s>}]^{1/\text{len}(y)}} \\
\log_2(\text{ppl}) &= -\frac{\log_2(\mathbb{P}[y \mathrel{|} \texttt{<s>}])}{\text{len}(y)} 
\end{align*}

## Text Classification

We have now studied two natural language processing tasks.
- **Tagging:** Given a text, assign each token a label
- **Language Modeling:** Given a text, predict the next token in the text

**Text Classification:** Given a text, assign the entire text a label.

## Classification Examples

- Spam filtering _(spam or ham?)_
- Sentiment analysis _(positive or negative?)_
- Hate speech detection _(harmful or safe?)_
- Language identification _(English, Swahili, or Quechua?)_
- Topic modeling _(business, politics, sports, or arts?)_
- Natural language inference _(entailment, contradiction, or neutral?)_
- Technically, language modeling _(a **ton** of possible labels)_

## Supervised vs. Unsupervised Learning

There are two ways of doing classification using machine learning.
- **Supervised Learning:** Each text in the training corpus has been manually annotated with a "gold-standard" label.
- **Unsupervised Learning (a.k.a. "clustering"):** Gold-standard labels are not available in the training corpus.

Language modeling is sometimes called "semi-supervised": gold-standard labels are available in the training corpus, but they do not have to be manually annotated.

## Bayesian Classification (Supervised)

Suppose $x$ is an email. We want to estimate:
- $\mathbb{P}[\text{spam} \mathrel{|} x]$
- $\mathbb{P}[\text{ham} \mathrel{|} x]$

Our predicted label is:
$$
\hat{y} = \begin{cases}
\text{spam}, &\mathbb{P}[\text{spam} \mathrel{|} x] > \mathbb{P}[\text{ham} \mathrel{|} x] \\
\text{ham}, &\text{otherwise.}
\end{cases}
$$

## Bayes's Theorem
$$\mathbb{P}[a \mathrel{|} b] = \frac{\mathbb{P}[b \mathrel{|} a] \mathbb{P}[a]}{\mathbb{P}[b]}$$

According to Bayes's Theorem,
$$\begin{align*}
    \mathbb{P}[\text{spam} \mathrel{|} x] &= \frac{\mathbb{P}[x \mathrel{|} \text{spam}]\mathbb{P}[\text{spam}]}{\mathbb{P}[x]} \\
    \mathbb{P}[\text{ham} \mathrel{|} x] &= \frac{\mathbb{P}[x \mathrel{|} \text{ham}]\mathbb{P}[\text{ham}]}{\mathbb{P}[x]}
\end{align*}$$

## Trick Question!

For classification, we don't actually have to calculate $\mathbb{P}[\text{spam} \mathrel{|} x]$ and $\mathbb{P}[\text{ham} \mathrel{|} x]$ _per se_; we just have to figure out which one is bigger.

**Observe:** 
$$
\begin{align*}
\mathbb{P}[\text{spam} \mathrel{|} x] > \mathbb{P}[\text{ham} \mathrel{|} x] &\iff \frac{\mathbb{P}[x \mathrel{|} \text{spam}]\mathbb{P}[\text{spam}]}{\mathbb{P}[x]} > \frac{\mathbb{P}[x \mathrel{|} \text{ham}]\mathbb{P}[\text{ham}]}{\mathbb{P}[x]} \\
&\iff\mathbb{P}[x \mathrel{|} \text{spam}]\mathbb{P}[\text{spam}] > \mathbb{P}[x \mathrel{|} \text{ham}]\mathbb{P}[\text{ham}]
\end{align*}$$

## Classification Score

To perform Bayesian classification, each label gets a **score**:
$$\begin{align*}
\text{score}(\text{spam}, x) &= \log_2(\mathbb{P}[x \mathrel{|} \text{spam}]) + \log_2(\mathbb{P}[\text{spam}]) \\
\text{score}(\text{ham}, x) &= \log_2(\mathbb{P}[x \mathrel{|} \text{ham}]) + \log_2(\mathbb{P}[\text{ham}])
\end{align*}$$

$\hat{y}$ is the label with the highest score.

## Bayesian Classification, Formalized

Let $C$ be a set of **labels** ("classes"). For a text $x$, a Bayesian classifier generates a **predicted label** $\hat{y} \in C$ as follows:

$$\hat{y} = \underset{c \in C}{\text{argmax}} \log_2(\mathbb{P}[x \mathrel{|} c]) + \log_2(\mathbb{P}[c])$$

where $\log_2(\mathbb{P}[x \mathrel{|} c])$ is estimated using a language model and $\log_2(\mathbb{P}[c])$ is the proportion of the training corpus with label $c$.

If $\log_2(\mathbb{P}[x \mathrel{|} c])$ is estimated using a unigram language model, then the classifier is called a **naive Bayes classifier**.
