## Warm-Up: Modularity

In computer programming, it is considered a good habit to write code that is divided into **"modular"** parts.

This means that if you write a function to do something, you should always use that function to do that thing.

```python
def vocab_size(corpus):
    """Calculates the vocab size of a corpus."""
    return len(set(corpus))

def lexical_diversity(corpus, percentage=False):
    """Calculates the lexical diversity (type–token ratio) of a corpus."""
    ld = vocab_size(corpus) / len(corpus)  # Calling the vocab_size function
    return ld * 100 if percentage else ld
```

## Recap: The Brown Corpus

The **Brown corpus** (Francis and Kučera, 1964) is a corpus of American English text in several **genres**.


```python
import nltk
from nltk.corpus import brown
nltk.download("brown")  # Download the corpus (you only have to do this once)
```

```python
# Load some parts of the Brown corpus
brown_news = brown.words(categories="news")
brown_fiction = brown.words(categories="fiction")
print(brown_news)
print(brown_fiction)
```

## Recap: FreqDists

We can use `FreqDist`s to find the counts for **every token type** in a corpus. 

```python
from nltk.probability import FreqDist
```

```python
FreqDist(brown_news)
```

```python
FreqDist(brown_news).most_common(5)
```

## FreqDist as a Data Type

`FreqDist` is actually a data type, and the `FreqDist(...)` function returns a `FreqDist` object!

```python
type(FreqDist(brown_news))
```

```python
# Save some FreqDists to variables
brown_news_dist = FreqDist(brown_news)
brown_fiction_dist = FreqDist(brown_fiction)
```

```python
print(brown_news_dist.most_common(5))
print(brown_fiction_dist.most_common(5))
```

## Recap: A Confusing Exercise

What are the most common **token lengths** in the `news` section of the Brown corpus?

First, we create a new "corpus" that contains the length of each token in the Brown corpus:

```python
brown_news_lengths = [len(w) for w in brown_news]
brown_news_lengths[:10]
```

Then, we use a `FreqDist` to count token frequencies in our corpus of token lengths:

```python
brown_news_lengths_dist = FreqDist(brown_news_lengths)
brown_news_lengths_dist
```

Finally, we use the `most_common` function to find the most common lengths:

```python
brown_news_lengths_dist.most_common(10)
```

## Recap: Dict Comprehension

Recall that we can create `dict`s from `lists` and `sets` using _dict comprehension_.

```python
# Example
capitals = {"UK": "London", "France": "Paris", "Germany": "Berlin", "Italy": "Rome"}

# What is this dict?
{capital: country for country, capital in capitals.items()}
```

## Dict Comprehension Exercise

Without using the `.values` function, how do you create a `list` containing the _values_ of a `dict`? For example, if

```python
x = {"a": 1, "b": 2, "c": 3}
```

then the values of `x` are `[1, 2, 3]`.

```python
print([v for _, v in x.items()])  # Darian's answer
print([i[1] for i in x.items()])  # Jun's answer (equally valid)
print([x[k] for k in x])          # Yet another possible answer
```
