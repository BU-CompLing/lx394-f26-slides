## Competence and Performance

Programming languages like TI-BASIC and Python consist of two components:

- a **specification** (i.e., grammar) that defines the syntax and semantics of the language
- an **interpreter** that parses and executes code.

## Python Specifications

In English syntax, we have rules like:

```
S  → DP VP
DP → Alice | Bob | Boston | D NP
VP → VP PP | V_trans DP
```

In [Python syntax](https://docs.python.org/3/reference/grammar.html), we have rules like:

```
comparison → sum eq
eq         → '==' sum
sum        → sum '+' term | term
```
