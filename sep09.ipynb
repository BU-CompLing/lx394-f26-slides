## Functions

A **function** is a piece of **reusable code** that maps _inputs_ to _outputs_.

```python
def frequency(corpus, word):    
    """ Calculates the frequency of a word in a corpus """
    return corpus.count(word) / len(corpus)  
```

```python
frequency(wiki_articles, "delve")
```

```python
frequency(chatgpt_articles, "delve")
```

```python
# How much more often does ChatGPT use "delve" than Wikipedia?
frequency(chatgpt_articles, "delve") / frequency(wiki_articles, "delve")
```
