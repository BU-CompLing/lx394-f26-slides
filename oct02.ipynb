## Recap: Efficiency

Constructing a `list` using a loop can be computationally inefficient.

```python
# Very inefficient
x = []
for i in range(5):
    x += [i * 2]  # Two new lists are created, x gets deleted
```

```python
# Somewhat efficient
x = []
for i in range(5):
    x.append(i * 2)  # x gets expanded, no new lists created
```

```python
# Very efficient
x = [i * 2 for i in range(5)]  # x is created using an optimized algorithm
```

## Warm-Up: Prime Numbers

We say that a number `x` is **prime** if: 
- `type(x) == int`
- `x > 1`
- if `type(i) == int` and `x % i == 0`, then either `i == 1` or `i == x`.

How do we check whether a number is prime?

```python
def is_prime(x: int) -> bool:
    for i in range(2, x):    
        if x % i == 0:
            return False
    return True
```

```python
is_prime(137)
```

## Hard Exercise (Trick Question)

How do we compute the first `n` primes using our `is_prime` function?

```python
def first_n_primes(n: int) -> list[int]:
    primes = []
    n_primes = 0
    for i in range(2, n * 2):
        if is_prime(i):
            primes.append(i)
            n_primes += 1
        if n_primes == n:
            return primes
```

```python
def first_n_primes(n: int) -> list[int]:
    i = 1
    first_n = []
    while len(first_n) < n:
        i += 1
        if is_prime(i):
            first_n.append(i)
    return first_n
```

## Solution Using a For Loop

This code works, but is very ugly.

```python
def fibonacci(n: int) -> int:
    fibs = [1, 1]
    for i in range(2, n + 1):
        fibs.append(fibs[-1] + fibs[-2])
    return fibs[-1]
```

## Solution Using Recursion

**Recursion** is when a function calls itself.

```python
def fibonacci(n: int) -> int:
    if n < 2:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)
```

## Recursion and Efficiency

The recursive version of `fib` is **extremely inefficient**. Why?

```python
def fib(n: int) -> int:
    if n < 2:
        return 1
    return fib(n - 1) + fib(n - 2)
```

How can we make `fib` more efficient?

## Memoization

One way to make `fib` more efficient is to **save** ("cache") **previously computed outputs**.

This is called **memoization**.

We can implement memoization in Python by creating a **class**.

## Classes, Objects, and Data Types

## Memoization Using a Class

```python
class Fibonacci:
    def __init__(self):
        self.cache = {0: 1, 1: 1}
        
    def fib(self, n: int) -> int:
        if n in self.cache:
            return self.cache[n]
        else:
            fib_n = self.fib(n - 1) + self.fib(n - 2)
            self.cache[n] = fib_n
            return fib_n
```

```python
fib = Fibonacci()
fib.fib(1300)
```
