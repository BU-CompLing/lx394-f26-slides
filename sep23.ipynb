## Warmup: Comparing Genres

Let's compare the token type frequences of the `news` section of the Brown Corpus with the `fiction` section.

````python
# Load some parts of the Brown corpus
import nltk
from nltk.corpus import brown
from nltk.probability import FreqDist

brown_news = brown.words(categories="news")
brown_fiction = brown.words(categories="fiction")
````

````python
brown_news_dist = FreqDist(brown_news)  # Count frequencies
brown_news_dist.most_common(10)          # Top 5 token types
````

````python
brown_fiction_dist = FreqDist(brown_fiction)
brown_fiction_dist.most_common(5)
````

## Recap: Comprehension

Recall that _comprehension_ lets us create new `list`s/`set`s/`dict`s from existing `list`s/`set`s/`dict`s.

````python
# What does range do?
list(range(10))
````

````python
# List comprehension: creating a list from a list
[x * 2 for x in range(10)]
````

````python
# Dict comprehension: creating a dict from a list
{x * 2: x for x in range(10)}
````

## Recap: Bound Variables

Recall that comprehensions require a _bound variable_ that represents the items in the original `list`. It doesn't matter which variable you use.

````python
[x * 2 for x in range(10)]
````

````python
[terrier * 2 for terrier in range(10)]
````

````python
# Dummy variable: use _ when the bound variable is not used
[5 for _ in range(10)]
````

## Comprehension over Dicts

Recall that there are a few different ways of doing comprehension over a `dict`.

````python
capitals = {"UK": "London", "France": "Paris", "Germany": "Berlin", "Italy": "Rome"}
````

````python
# Comprehension with keys
[capital.lower() for capital in capitals]
````

````python
# Comprehension with values
[capital.lower() for capital in capitals.values()]
````

````python
# Comprehension with keys and values
[capital + ", " + country for country, capital in capitals.items()]
````

## Many Different Solutions

There are often many different ways of doing the same thing.

For example, here are two equally valid ways to "invert" the `capitals` `dict`.

````python
{capital: country for country, capital in capitals.items()}
````

````python
{capitals[country]: country for country in capitals}
````

## Preprocessing

Sometimes we want to **preprocess** a corpus before studying it. For example, we might want to remove punctuation marks and **stopwords**.

````python
# Punctuation marks
import string
set(string.punctuation)
````

````python
# Stopwords
from nltk.corpus import stopwords
nltk.download("stopwords")
stopwords.words("english")
````

## Exercise

How would we write a function to filter out punctuation and stopwords from a corpus?

````python
def preprocess(corpus):
    """Removes punctuation and stopwords from a corpus"""
    return  [w for w in corpus if w not in string.punctuation and 
             w.lower() not in stopwords.words("english")]
````

````python
brown_news_dist = FreqDist(preprocess(brown_news))
brown_fiction_dist = FreqDist(preprocess(brown_fiction))
````

````python
brown_news_dist.most_common(10)
````

````python
brown_fiction_dist.most_common(10)
````
