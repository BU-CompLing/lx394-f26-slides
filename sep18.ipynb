NLTK contains a number of other corpora that have been pre-prepared. For example, let's download and load the **Brown corpus** (Francis and Kučera, 1964), a corpus of American English text in several **genres**.


```python
# Download the corpus (you only have to do this once)
import nltk
nltk.download("brown")
```

## Using the Brown Corpus

```python
# Load the corpus
from nltk.corpus import brown
brown.categories()
```

```python
brown.words(categories="news")
```

```python
type(brown.words(categories="news"))
```

```python
len(brown.words(categories="news"))
```

## Most Common Token Types

What are the most common token types (i.e., lexical items) in the `news` section of the Brown Corpus?

We can find this using a `FreqDist`.

```python
from nltk.probability import FreqDist
```

```python
dist = FreqDist(brown.words(categories="news"))
dist["Boston"]
```

```python
dist.most_common(10)
```

## Exercise

What are the most common **token lengths** in the `news` section of the Brown corpus?

First, we create a new "corpus" that contains the length of each token in the Brown corpus:

```python
lengths = [len(w) for w in brown.words(categories="news")]
print(brown.words(categories="news")[:10])
lengths[:10]
```

Then, we use a `FreqDist` to count token frequencies in our corpus of token lengths:

```python
lengths_dist = FreqDist(lengths)
lengths_dist
```

## Exercise

What are the most common **token lengths** in the `news` section of the Brown corpus?

Finally, we use the `most_common` function to find the most common lengths:

```python
lengths_dist.most_common(15)
```

## Recap: Dicts

A `FreqDist` is really a special type of `dict`. A `dict` (short for "dictionary") is like a `list`, but it's not necessarily indexed by `int`s.

In other programming languages, `dicts` are sometimes called **mappings** or **hashmaps**.

```python
x = {"a": 1, "b": 2, "c": 3}
x["a"]
```

```python
len(x)
```

## Recap: Comprehension

Recall that _comprehension_ lets us create a `list`/`set`/`dict` from an existing `list`/`set`/`dict`.

```python
x.keys()
```

```python
# Take each k in x, add "1" to the end, and put it in a list
[k + "1" for k in x.keys()]
```

```python
# This does the same thing
[k + "1" for k in x]
```

```python
# List of key-value pairs
x.items()
```

```python
# Give me a dict where for each (k, v), k is mapped to v + 1
{k: v + 1 for k, v in x.items()}
```
